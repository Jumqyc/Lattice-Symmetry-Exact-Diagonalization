```python
from base import Lattice
import numpy as np
import matplotlib.pyplot as plt

from itertools import product

square = Lattice(
    cell_info={
        "A": np.array([0, 0])},
    supercell_info=[
        (np.array([1, 0]), 5),
        (np.array([0, 1]), 5)]
)

ax = square.plot()
plt.show()


# transfer matrix
# rotation by 90 degrees
rotation_90 = np.array([[0, -1], [1, 0]])
print("Permutation for 90 degree rotation:", square.get_permutation_pt(rotation_90))
x_flip = np.array([[-1, 0], [0, 1]])
print("Permutation for x flip:", square.get_permutation_pt(x_flip))
diagonal_flip = np.array([[0, 1], [1, 0]])
print("Permutation for diagonal flip:", square.get_permutation_pt(diagonal_flip))

# translation by x direction
print("Permutation for translation in x direction:", square.get_permutation_tr(0))
print("Permutation for translation in y direction:", square.get_permutation_tr(1))
```

```python
triangular = Lattice(
    cell_info={
        "A": np.array([0, 0]),
    },
    supercell_info=[
        (np.array([1, 0]), 5),
        (np.array([1/2, np.sqrt(3)/2]), 5)]
)
triangular.center_at(4)

ax = triangular.plot()
plt.show()

rotation_60 = np.array([[1/2, -np.sqrt(3)/2], [np.sqrt(3)/2, 1/2]])
print("Permutation for 60 degree rotation:", triangular.get_permutation_pt(rotation_60))
print("Permutation for x flip", triangular.get_permutation_pt(x_flip))
```

```python
hexagonal = Lattice(
    cell_info = {
        'A': np.array([0, 0]),
        'B': np.array([1, 0])
    },
    supercell_info = [
        (np.array([1.5, np.sqrt(3)/2]), 4),
        (np.array([1.5, -np.sqrt(3)/2]),4)
    ]
)
hexagonal.center_at(np.array([2,0]))

ax = hexagonal.plot()
# rotate by 60 degrees
plt.show()
print("Permutation for 60 degree rotation:", hexagonal.get_permutation_pt(rotation_60))
print("Permutation for x flip", hexagonal.get_permutation_pt(x_flip))
```
