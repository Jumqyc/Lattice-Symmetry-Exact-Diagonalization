```python
import numpy as np
import matplotlib.pyplot as plt
from lattice_action import AffineAction, Lattice

x_flip = np.diag([-1.0, 1.0])
```

```python
square = Lattice(
    cell_info={
        ('A', 0): np.array([0.0, 0.0])},
    supercell_info=[
        (np.array([1.0, 0.0]), 3),
        (np.array([0.0, 1.0]), 3)
    ]
)

ax = square.plot()
plt.show
```

```python
triangular = Lattice(
    cell_info={
        ('A', 0): np.array([0, 0]),
    },
    supercell_info=[
        (np.array([1, 0]), 5),
        (np.array([1/2, np.sqrt(3)/2]), 5)]
)
triangular.set_center_at(4)

ax = triangular.plot()
plt.show()

rotation_60 = np.array([[1/2, -np.sqrt(3)/2], [np.sqrt(3)/2, 1/2]])
print("Permutation for 60 degree rotation:",
      AffineAction.from_matrix(rotation_60, 2).site_map(triangular))
print("Permutation for x flip:",
      AffineAction.from_matrix(x_flip, 2).site_map(triangular))
```

```python
a1 = np.array([1.0, 0.0])
a2 = np.array([0.5, np.sqrt(3) / 2])
bond = (a1 + a2) / 3
centre = (a1 + a2) / 2          # a hexagon centre: the origin of the 60 degree rotation

honeycomb = Lattice(
    cell_info={
        ('C', 0): -bond / 2 - centre,
        ('C', 1): +bond / 2 - centre},
    supercell_info=[(a1, 6), (a2, 6)]
)

ax = honeycomb.plot()
plt.show()
print("Permutation for 60 degree rotation:",
      AffineAction.from_matrix(rotation_60, 2).site_map(honeycomb))
print("Permutation for x flip:",
      AffineAction.from_matrix(x_flip, 2).site_map(honeycomb))
```
